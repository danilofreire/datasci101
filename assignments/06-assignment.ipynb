### Question 2: Data Drift after COVID

Imagine a restaurant recommendation app trained on user reviews and ratings collected between 2017 and 2019. The app learns which restaurants people like based on cuisine type, price range, location, and review sentiment. Then COVID hits in 2020, and the world changes.

Lecture 14 described three types of drift:

- **Concept drift**: the *relationship* between inputs and outputs changes (P(Y|X) shifts)
- **Data drift**: the *inputs themselves* change (P(X) shifts)
- **Label drift**: the *distribution of outcomes* changes (P(Y) shifts)

**Your task:**

1. Describe one specific example of **concept drift** that could affect this restaurant app after COVID. (Hint: did the meaning of a "good restaurant experience" change?)

2. Describe one specific example of **data drift** — how did the types of reviews or the kinds of users submitting reviews change?

3. Describe one specific example of **label drift** — how did the distribution of star ratings shift?

4. For each type of drift, propose one way the company could detect that it's happening. Be concrete: what data would you compare, and what pattern would you look for?

### Bonus Question: Write a Mini-Datasheet

Lecture 15 introduced **datasheets for datasets**, documentation that describes where a dataset came from, what's in it, and how it should (and shouldn't) be used. The framework comes from Gebru et al. (2018) and has 7 sections.

Here is a dataset you need to document:

> **Atlanta Restaurant Reviews Dataset**: 1,000 Yelp reviews of restaurants in the Atlanta metro area, scraped in March 2024. Each entry contains the review text, a star rating (1-5), the restaurant name, the restaurant's cuisine type, and the date of the review. The reviews were collected by a graduate student at Emory University for a class project on sentiment analysis. No usernames or profile photos were included in the scrape.

**Your task:**

Fill out each of the 7 datasheet sections below. You'll need to think carefully about what information is available and what's missing. Some sections will require you to make reasonable assumptions — state those assumptions shortly but clearly.

1. **Motivation**: Why was this dataset created? Who created it and who funded it?
2. **Composition**: What types of data are included? How many instances? Is there sensitive information?
3. **Collection**: How was the data collected? Was consent obtained from the reviewers?
4. **Preprocessing**: Was the data cleaned or filtered? Were any instances removed?
5. **Uses**: What tasks has it been used for? What should it NOT be used for?
6. **Distribution**: How is the dataset shared? Is it versioned?
7. **Maintenance**: Who is responsible for updates? Is there a way to report issues?

After filling out the datasheet, identify **2 ethical concerns** with this dataset that the datasheet helps reveal.
