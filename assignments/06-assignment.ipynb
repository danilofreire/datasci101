### Question 2: Drift after COVID

Imagine a restaurant recommendation app trained on reviews and ratings
from 2017 to 2019. It learns which restaurants people like from cuisine
type, price range, location and review sentiment. Then COVID arrives in
2020.

Lecture 14 described three types of drift:

- **Concept drift**: the *relationship* between inputs and outputs
  changes (P(Y\|X) shifts)
- **Data drift**: the *inputs themselves* change (P(X) shifts)
- **Label drift**: the *distribution of outcomes* changes (P(Y) shifts)

**Your task:**

1.  Give one specific example of **concept drift** that could affect the
    app after COVID. (Hint: did the meaning of a “good restaurant
    experience” change?)

2.  Give one specific example of **data drift**. How did the reviews, or
    the people writing them, change?

3.  Give one specific example of **label drift**. How did the
    distribution of star ratings change?

4.  For each type of drift, propose one way the company could detect it.
    Say what data you would compare and what pattern you would look for.
