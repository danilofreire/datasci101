### Question 1: The Inference Problem

Lecture 19 showed that AI can predict sensitive attributes from seemingly innocuous data. Facebook likes predict political views and sexuality. Purchase history predicts pregnancy (the Target case). Location data reveals religion, home address, and health conditions. The lecture called this the "inference problem": you can control what you share, but you cannot control what can be inferred from what you share.

**Your task:**

1. You install a free weather app that tracks your location. List **3 pieces of sensitive information** that could be inferred from your location data alone, not information you typed in, but things that can be figured out from where you go and when. For each, explain the inference chain (e.g., "regularly visits X on Sundays → likely Y").

2. Traditional privacy protections centre on controlling what you share: don't give out your medical records, don't post your address online, etc. In 3–4 sentences, explain why the inference problem makes these traditional protections less effective. What changes when sensitive information can be derived from non-sensitive data?

3. Propose **one technical rule** and **one policy rule** that would address the inference problem. For each, explain what it would do and why it might be hard to implement.

### Question 2: "Publicly Available" Does Not Mean "Consented to AI Training"

Lecture 19 emphasised that most people don't know their data is in AI training sets, and that data being publicly accessible does not mean the person agreed to have it used for model training. **Your task:**

1. Give **3 specific examples** of data that is "publicly available" but where the person likely never imagined it would be used to train an AI model. For each, explain why "it was public" is not a sufficient justification for including it in training data.

A social media company updates its terms of service to say: *"By posting on our platform, you consent to your content being used to train AI models."* You already have 5 years of posts on the platform. In 3–4 sentences, argue whether retroactive consent (applying new terms to old posts you made under different terms) is ethically valid. Consider: did you have a meaningful choice when you originally posted?

3. The EU's GDPR requires consent to be "freely given, specific, and informed." Could a social media platform ever meet this standard for AI training consent? Why or why not? Write 3–4 sentences.

### Question 3: Synthetic Data: Too Good to Be True?

Lecture 19 described synthetic data: artificially generated data that preserves the statistical patterns of real data without containing actual personal information. Healthcare organisations use synthetic patient records for research. Financial institutions test fraud detection on synthetic transactions. The appeal is clear: you get useful data without the privacy risk of sharing real records.

But the lecture also warned about limitations: synthetic data may not capture rare cases well, and there is a risk of "overfitting" to original data (the generator memorises specific real individuals).

**Your task:**

1. A hospital has 10,000 real patient records. They generate 10,000 synthetic records and give them to a research team. Explain in 3–4 sentences why this is better for privacy than simply "anonymising" the real records (e.g., replacing names with ID numbers, removing dates of birth). What does synthetic data protect against that basic anonymisation does not?

2. The lecture warned that synthetic data "may not capture rare cases well." Give a **specific healthcare example** where this limitation could cause harm. Describe the rare case, explain why the synthetic data might miss it, and identify who would be harmed.

3. There is a risk that the synthetic data generator "memorises" specific real patients, that is, producing synthetic records that are too close to real ones. How is this similar to the LLM memorisation problem discussed earlier in the lecture (where models regurgitate verbatim training data)? How is it different? Is one more dangerous than the other? Write 3–4 sentences.

### Question 4: Facial Recognition in Public Spaces

The EU AI Act (Lecture 18) banned real-time biometric identification in public spaces, with exceptions for serious crimes. China deploys facial recognition extensively in public, including for surveillance. The US has no federal ban; the rules depend on state and city laws, and many jurisdictions have no restrictions at all.

Lecture 19 discussed Clearview AI, which scraped billions of photos from social media to build a facial recognition database and sold access to law enforcement and private buyers.

### Question 5: The Right to Be Forgotten in the Age of LLMs

GDPR gives individuals the "right to erasure", that is, the right to demand that an organisation delete your personal data. But Lecture 19 noted that removing one person's data from a trained AI model is technically very difficult. The lecture described this as one of the key tensions between privacy law and AI: the right exists on paper, but technology may not be able to comply.

**Your task:**

1. Explain in plain language why deleting a row from a database is simple, but "deleting" someone's data from a neural network is not. What makes the two cases fundamentally different? Write 3–4 sentences.

2. Some researchers are working on "machine unlearning", techniques to make a model forget specific training examples without retraining the entire model from scratch. Even if this works perfectly, does it fully solve the privacy problem? Consider: what if the model already generated outputs based on your data before the unlearning happened? What if those outputs were used to make decisions about other people? Write 3–4 sentences.

3. Should regulators require that LLM providers demonstrate the ability to remove individual data upon request, even if it is technically expensive? Or should the right to erasure be limited for AI models, recognising the technical constraints? Take a clear position in 4–5 sentences. Address at least one objection someone from the opposite side might raise.
