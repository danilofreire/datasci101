## The case

A company has offered Emory a remote proctoring system for online exams.
We will call it **Invigil**. During an exam, Invigil takes a photograph
through the student’s webcam every three seconds and records audio for
the whole session. Software scores each exam and flags some for review
by the Honor Council. The brochure says the system is **96% accurate**.

In a pilot of **2,000 exams** at another university, an independent
audit found **40** cases of cheating. Invigil flagged **80** exams. The
Honor Council confirmed cheating in **20** of them. So three in four
flags were honest students, and half the cheaters were never flagged.

The provost has asked your committee to assess Invigil before the
university signs anything. Every question below uses this case.
