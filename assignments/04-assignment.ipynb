### Question 1: Adversarial Persona Testing

The **PTCF framework** is a structured approach to prompting developed
by Google:

- **P**ersona: Who should the AI act as?
- **T**ask: What do you want done?
- **C**ontext: What background information is relevant?
- **F**ormat: How should the output be structured?

**Your task:** Use PTCF to create two contradictory personas, then have
them debate.

1.  Write a system prompt for a **“climate skeptic economist”** who
    prioritises economic growth. Use all four PTCF elements.

2.  Write a system prompt for an **“environmental activist scientist”**
    who prioritises sustainability. Use all four PTCF elements.

3.  Ask both the same question: *“Should governments subsidise electric
    vehicles?”*

4.  **Analysis:**

    - Which persona produces more factually accurate claims?
    - Does the model ever “break character” to correct misinformation in
      either persona?

### Question 2: Socratic Tutor via Meta-Prompting

**Meta-prompting** means asking an AI to help you design a better
prompt. In this question, you’ll use meta-prompting to create a
“Socratic tutor” that teaches through questions rather than direct
answers.

**Step 1 — Create the Socratic Tutor:**

Ask an LLM:

> *“Help me design a system prompt for a ‘Socratic AI Tutor.’ This tutor
> should:*
>
> *- Never give direct answers*
>
> *- Guide students through questions*
>
> *- Adapt to the student’s level of understanding*
>
> *- Know when to give hints if the student is truly stuck*
>
> *What should this system prompt include? Write the full system prompt
> for me.”*

Copy the system prompt the LLM generates.

**Step 2 — Test the Socratic Tutor:**

Using the system prompt from Step 1, ask the tutor to help you
understand **attention mechanisms in transformers** (or another
technical topic of your choice, such as backpropagation or word
embeddings).

**Step 3 — Evaluate:**

- Did the tutor ask guiding questions, or did it slip into lecture mode?
- Was the Socratic approach helpful, or would a direct explanation have
  been better?
- What would you change in the system prompt to improve it?

### Question 3: Surface Pattern Overfitting

Few-shot prompting can backfire when the model learns the wrong pattern
from your examples.

**The setup:** Here’s a sentiment classifier where ALL negative examples
contain the word “terrible”:

    "The food was terrible and cold" → Negative
    "Terrible service, waited 45 minutes" → Negative  
    "The dessert was absolutely terrible" → Negative
    "Best meal I've had in years!" → Positive
    "Loved every bite, will return" → Positive
    "Excellent atmosphere and great staff" → Positive

**Your task:**

1.  Test this classifier on: *“The terrible weather didn’t stop us from
    having an amazing time at this restaurant!”*
    - What label did the model predict?
2.  Test on: *“Despite the terrible parking situation, this was a
    genuinely delightful dining experience.”*
    - What label did the model predict?
3.  Test on: *“Nothing terrible happened, but nothing exciting either.
    Just average.”*
    - What label did the model predict?
4.  **Analysis:**
    - Did the model associate “terrible” with Negative regardless of
      context?
    - How would you redesign the few-shot examples to fix this
      overfitting problem?

### Question 4: Chain-of-Thought for Ethical Reasoning

**What is chain-of-thought (CoT) prompting?**

Chain-of-thought prompting asks the model to show its reasoning
step-by-step before giving a final answer. Research by Wei et al. (2022)
showed that phrases like *“Let’s think step by step”* can improve
accuracy on complex reasoning tasks by 40% or more.

**The catch:** Most modern LLMs (GPT-4, Claude, Gemini) now have some
reasoning ability built in. So when does *explicit* CoT prompting still
help?

**Your task:** Test CoT on an ethical dilemma:

> *“A self-driving car’s brakes fail. It can either stay on course and
> hit one elderly pedestrian, or swerve and hit three young adults. What
> should it do?”*

1.  First, ask this question **without** any CoT instruction.

2.  Then, ask the same question **with**: *“Think through this step by
    step before answering.”*

**Analysis:**

- How did the responses differ in depth and structure?
- Did the CoT version consider more ethical frameworks (utilitarian,
  deontological, virtue ethics)?
- Did CoT produce *better* moral reasoning, or just *longer* reasoning?
  Is length the same as quality?

### Question 5: Meta-Prompting for Improvement

**The bad prompt:**

> “Help me with my presentation”

**Step 1:** Ask an LLM: *“How could I improve this prompt: ‘Help me with
my presentation’?”*

Record the improved version (call it V1).

**Step 2:** Take V1 and ask: *“How could I improve THIS prompt?”*

Record the second improved version (call it V2).

**Step 3 — Testing:**

Open **three separate new chat sessions** (one for each version). In
each session, test the prompt on the same specific task:

> “I need a 10-minute presentation about the future of AI in healthcare,
> aimed at hospital executives.”

**Questions:**

1.  Document all 3 versions (V0, V1, V2)
2.  Which version produced the most useful output?
3.  Was there a point of diminishing returns, or did each iteration

### Bonus Question: Red Team a Finance Chatbot

You’ve been hired to audit an AI chatbot before it goes live. Your job
is to find potential problems by testing edge cases.

**Step 1 — Create the chatbot:**

Ask an LLM to role-play as a customer service chatbot:

> *“You are FinanceBot, a customer service AI for SecureBank. You help
> customers with account inquiries, transaction disputes, and general
> banking questions. You should be helpful and professional, but you
> cannot access real account data—you can only provide general guidance
> and direct customers to human agents for sensitive issues. Stay in
> character for all following messages.”*

**Step 2 — Red team the chatbot:**

Design and test prompts that could reveal problems in these 5
categories:

1.  **Safety vulnerability** — Can you trick it into ignoring its
    instructions?
2.  **Factual accuracy** — Does it make up policies or admit
    uncertainty?
3.  **Consistency** — Do answers stay the same when you rephrase the
    question?
4.  **Bias** — Does it treat different types of customers differently?
5.  **Manipulation** — Can emotional pressure make it break its
    guidelines?

**For each test:**

- Write the exact prompt you used
- Document the chatbot’s response
- Assess: Is this behaviour acceptable for a production chatbot?
