### Question 1: Adversarial Persona Testing

The **PTCF framework** is a structured approach to prompting developed by Google:

- **P**ersona: Who should the AI act as?
- **T**ask: What do you want done?
- **C**ontext: What background information is relevant?
- **F**ormat: How should the output be structured?

**Your task:** Use PTCF to create two contradictory personas, then have them debate.

2. Write a system prompt for an **"environmental activist scientist"** who prioritises sustainability. Use all four PTCF elements.

3. Ask both the same question: *"Should governments subsidise electric vehicles?"*

4. **Analysis:** 
   - Which persona produces more factually accurate claims? 
   - Does the model ever "break character" to correct misinformation in either persona?

### Question 4: Chain-of-Thought for Ethical Reasoning

**What is chain-of-thought (CoT) prompting?**

Chain-of-thought prompting asks the model to show its reasoning step-by-step before giving a final answer. Research by Wei et al. (2022) showed that phrases like *"Let's think step by step"* can improve accuracy on complex reasoning tasks by 40% or more.

**The catch:** Most modern LLMs (GPT-4, Claude, Gemini) now have some reasoning ability built in. So when does *explicit* CoT prompting still help?

**Your task:** Test CoT on an ethical dilemma:

> *"A self-driving car's brakes fail. It can either stay on course and hit one elderly pedestrian, or swerve and hit three young adults. What should it do?"*

1. First, ask this question **without** any CoT instruction.

2. Then, ask the same question **with**: *"Think through this step by step before answering."*

**Analysis:**

- How did the responses differ in depth and structure?
- Did the CoT version consider more ethical frameworks (utilitarian, deontological, virtue ethics)?
- Did CoT produce *better* moral reasoning, or just *longer* reasoning? Is length the same as quality?
