### Question 1: Training Data and the Interpolation Problem

Lecture 10 explained that for rare topics, LLMs “fill in gaps” using
patterns from similar-but-different topics. This is called
**interpolation** and it’s where confident nonsense comes from.

For this assignment, we’ll use [chat.z.ai](https://chat.z.ai/) because
you can turn off both internet search and deep thinking. This is
important: if the model can search the web, it won’t hallucinate as
much, and we want to see hallucinations in action.

**Your task:**

1.  Go to [chat.z.ai](https://chat.z.ai/). **Turn off both “Search” and
    “Think” by clicking the toggles below the chat box**. We want to
    test the model’s raw knowledge, without help.
2.  Ask: *“Tell me about the 2019 general election in Liechtenstein. Who
    won, what were the main issues, and what was the voter turnout?”*
3.  Copy the response.
4.  Now verify on Wikipedia: [Liechtenstein
    elections](https://en.wikipedia.org/wiki/Elections_in_Liechtenstein).
    Check: Did the LLM invent party names? Did it get the year right?
    Are the issues plausible but fabricated?
5.  Now ask: *“Tell me about the 1999 US presidential election. Who won
    and what was the voter turnout?”*
6.  Try to verify this answer on Wikipedia: [US presidential
    elections](https://en.wikipedia.org/wiki/United_States_presidential_election).

**Analysis:**

- Did the model answer confidently about an election that never
  happened?
- How confident did the model *sound* in each response? Did it hedge or
  express uncertainty, or did it present fabricated information as fact?
- Why does the amount of training data on a topic affect hallucination
