### Bonus question: Is hallucination lying?

A lie needs an intent to deceive. A model has no intentions, yet it
produces false statements in the same confident voice it uses for true
ones. The philosopher Harry Frankfurt called speech that is indifferent
to truth, aimed only at sounding right, “bullshit”, and argued it is
more corrosive than lying, because the liar at least keeps track of the
truth. [Hicks, Humphries and Slater
(2024)](https://doi.org/10.1007/s10676-024-09775-5) apply this to
ChatGPT in a short, open-access paper: they argue that “hallucination”
is the wrong word, and “bullshit” is the accurate one.

Write 150-250 words:

- Do you agree that a model’s false statements are closer to bullshit
  than to lying? Use one example from this assignment to make your case
- Suppose the industry adopted the word. What would change for users,
  for the companies that sell these systems, and for how you use them in
  your own work?
- Frankfurt says the bullshitter does not care whether what they say is
  true. The people who train models clearly care. Does that rescue the
  term “hallucination”, or does it just move the indifference somewhere
  else?
