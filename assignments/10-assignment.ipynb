### Question 1: The Skill Compression Effect

Several studies discussed in Lecture 20 found that AI tools boost
productivity, but the gains are uneven. The BCG consulting study (2023)
found a 40% performance improvement on applicable tasks. Brynjolfsson et
al. (2025) found a 14% productivity increase in customer service, with
the biggest gains going to **novice workers**. GitHub’s Copilot study
found 55% faster task completion for programmers. Across these studies,
a pattern emerges: AI **compresses the skill distribution**, helping
beginners more than experts. In some cases, experts actually perform
*worse* with AI assistance.

**Your task:**

    judgment) or down (their edge has shrunk)? Argue for one scenario
    and explain your reasoning in 4–5 sentences.

    fewer juniors** even though juniors benefit most from AI. If AI

### Question 2: Augmentation vs Replacement — Why It Matters Who Decides

Daron Acemoglu, an MIT economist who won the 2024 Nobel Prize, argues
that current AI development is “biased toward replacement”, that is,
automating workers out of tasks rather than making them better at new
ones. The lecture distinguished between AI that **augments** workers
(makes them more productive at higher-value tasks) and AI that
**replaces** them (does the task instead). Acemoglu’s concern is that
companies have stronger financial incentives to replace than to augment.

**Your task:**

1.  Give **one specific example** of AI being used to augment a worker
    (making them better at their job) and **one specific example** of AI
    being used to replace a worker (doing the job instead). For each,
    explain which tasks shift and who benefits. Write 2–3 sentences per
    example.

2.  Even when augmentation would produce better long-term outcomes
    (higher-quality work, more innovation), companies often choose
    replacement. In 3–4 sentences, explain why. Think about what shows
    up on a quarterly earnings report and what doesn’t.

    incentive, that would nudge companies toward augmentation rather
    than replacement. Explain how it would work in practice and identify
    **one realistic objection** someone might raise against it. Write
