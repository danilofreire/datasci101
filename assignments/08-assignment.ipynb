### Question 1: Clinical Notes Describe Patients Differently by Race

Research has found that doctors write about patients differently
depending on the patient’s race and gender. Studies by Sun et al. (2022)
and Beach et al. (2021) showed that clinical notes are more likely to
use words like “non-compliant,” “refuses,” or “aggressive” for Black
patients, and words like “pleasant,” “cooperative,” or “well-appearing”
for white patients, even when the patients’ actual behaviour and
conditions were similar.

Lecture 16 discussed how LLMs are now used to read and summarise
clinical notes in systems like Epic’s MyChart. These models learn from

    and how a doctor reading the summary could be misled.

    explain why this makes bias in clinical notes harder to detect and

### Question 4: The EU AI Act Risk Pyramid Has a Grey Zone

The EU AI Act classifies AI systems into four tiers: **Prohibited**
(banned outright), **High Risk** (strict requirements), **Limited Risk**
(transparency obligations), and **Minimal Risk** (no requirements). In
Lecture 17, the examples were clear-cut: a spam filter is minimal risk,
a hiring algorithm is high risk, and government social scoring is
prohibited.

But many real systems don’t fit neatly into one category.

**Here is the scenario:** A university deploys an AI system that reads
students’ essays and flags those it suspects were written by AI. Flagged
essays go to a professor, who reviews them before deciding on any
academic integrity action. The student is not told their essay was
flagged unless the professor decides to pursue the case.

**Your task:**

1.  **Argue that this system is high risk.** Consider: the system
    affects educational access and academic standing. A false positive
    (flagging a human-written essay as AI-generated) could lead to
    disciplinary proceedings. Write 3–4 sentences making the strongest
    case for high-risk classification.

2.  **Now argue that it is limited risk.** Consider: the system is a
    detection tool, not a decision-maker. A human professor reviews
    every flagged case before any action is taken. Write 3–4 sentences
    making the strongest case for limited-risk classification.

3.  Which classification do you find more convincing? In your answer
    (4–5 sentences), address **both** of these questions:

    - Does the presence of a human reviewer actually make the system
      less risky, or does it just shift responsibility?
    - What matters more for classification: the technology itself, or
