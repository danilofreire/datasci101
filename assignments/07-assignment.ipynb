### Question 2: The Proxy Variable Hunt

Lecture 16 explained **measurement bias**: when a measurable proxy is
used as a stand-in for something you actually care about, but the proxy
doesn’t work equally for everyone. The classic example is using ZIP
codes in loan decisions. They correlate with race because of historical
housing discrimination, so a “race-neutral” variable ends up encoding
race.

Here is a fictional dataset used by a bank to predict whether someone

### Question 3: The Impossibility Theorem: Pick Your Poison

Lecture 16 introduced the **impossibility theorem**: if two groups have
different base rates (e.g., different loan repayment rates), you cannot
satisfy all fairness criteria at the same time. You have to choose, and
every choice harms someone.

Here is the scenario. A city government uses an algorithm to decide
which arrested individuals can be released on bail before trial. The
algorithm predicts the probability that each person will show up for
their court date.

**The data shows:**

- **Group A**: 80% show up for their court date
- **Group B**: 60% show up for their court date (due to factors like
  lack of transport, inflexible jobs, and childcare, not criminal
  intent)

The algorithm assigns a risk score from 0 to 100. The city must set a
threshold: anyone above the threshold is released on bail.

**Three possible approaches:**

| Approach | Rule | Consequence |
|-------------------------|----------------|--------------------------------|
| **Same threshold** | Everyone with score \> 50 is released | More Group B detained (because their scores are lower on average) |
| **Equal release rates** | Adjust thresholds so 60% of each group is released | More Group A members who would have skipped court are released; public safety concern |
| **Equal false detention rates** | Ensure the same percentage of people who *would have* shown up are wrongly detained in both groups | Different release rates; Group B gets a lower threshold, which some call “preferential treatment” |

**Your task:**

1.  For **each approach**, identify who is harmed and how. Be specific:
    don’t just say “it’s unfair.” Describe a person who suffers under
    each rule.

2.  Choose the approach you think is least bad. Write at least 3-4
    sentences defending your choice. You must address at least one
    objection someone could raise against your choice.

3.  The reasons Group B has a lower show-up rate (transport, jobs,
    childcare) are not about criminal intent. Does this change which
    approach you’d pick? Should the algorithm account for *why* someone
    is risky, or just *whether* they are? Explain in 3–4 sentences.

### Question 4: Breaking a Feedback Loop

Lecture 16 described **feedback loops**: when an algorithm’s predictions
influence the data it will be trained on in the future. The predictive
policing example showed how this works: the algorithm sends police to
certain neighbourhoods, more arrests happen there, and the new arrest
data “confirms” the algorithm’s prediction, even if crime rates are
actually similar across areas.

> **TrustMod** is a social media platform that uses an AI content

> next version of the model.
>
> Over the past year, TrustMod’s moderation team has noticed a pattern:
> the system flags posts written in informal or regional dialects of
> English at roughly **twice the rate** of posts written in standard

    cycle? Propose **two different interventions**: one technical (e.g.,

3.  A TrustMod engineer says: *“The model isn’t biased. It’s just

### Bonus Question: The Simpson’s Paradox Trap

Lecture 16 discussed **aggregation bias**, which occurs when treating
diverse groups as one hides important differences. A related statistical
phenomenon is **Simpson’s paradox**: a trend that appears in aggregated
data can reverse when you look at subgroups.

Here is a real-sounding (but fictional) scenario:

> A hospital is testing two treatments for kidney stones: **Treatment
> A** (a new, less invasive procedure) and **Treatment B** (traditional
> surgery). They collect data on 700 patients.

**Overall results:**

| Treatment | Patients Treated | Patients Recovered | Recovery Rate |
|-----------|------------------|--------------------|---------------|
| A         | 350              | 273                | 78%           |
| B         | 350              | 289                | 83%           |

Treatment B looks better. But let’s break it down by stone size:

**Small stones:**

| Treatment | Patients | Recovered | Recovery Rate |
|-----------|----------|-----------|---------------|
| A         | 87       | 81        | 93%           |
| B         | 270      | 234       | 87%           |

**Large stones:**

| Treatment | Patients | Recovered | Recovery Rate |
|-----------|----------|-----------|---------------|
| A         | 263      | 192       | 73%           |
| B         | 80       | 55        | 69%           |

**Your task:**

1.  Confirm the paradox: Treatment B wins overall (83% vs 78%), but
    Treatment A wins for *both* small stones (93% vs 87%) and large
    stones (73% vs 69%). In 2–3 sentences, explain how this is possible.
    (Hint: look at how patients were distributed across groups.)

2.  A hospital administrator looks only at the overall table and
    announces: *“Treatment B is better. We’re switching all patients to
    Treatment B.”* Explain why this decision is wrong and what harm it
    could cause. Be specific.

3.  Connect this to **Lecture 16**: which type of bias from the taxonomy
    does Simpson’s paradox illustrate? Why does Lecture 16 argue that
    **disaggregated evaluation** (breaking results down by group) is
    necessary, not optional?

4.  Propose a rule the hospital could adopt to prevent this kind of
    mistake in future decisions. Your rule should be concrete and
    actionable, not just “be careful with data.”
