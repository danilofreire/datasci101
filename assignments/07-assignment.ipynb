Lecture 15 explained **measurement bias**: when a measurable proxy is used as a stand-in for something you actually care about, but the proxy doesn't work equally for everyone. ### Question 3: The Impossibility Theorem: Pick Your Poison

Lecture 15 introduced the **impossibility theorem**: if two groups have different base rates (e.g., different loan repayment rates), you cannot satisfy all fairness criteria at the same time. You have to choose, and every choice harms someone.

Here is the scenario. A city government uses an algorithm to decide which arrested individuals can be released on bail before trial. The algorithm predicts the probability that each person will show up for their court date.

**The data shows:**

- **Group A**: 80% show up for their court date
- **Group B**: 60% show up for their court date (due to factors like lack of transport, inflexible jobs, and childcare, not criminal intent)

The algorithm assigns a risk score from 0 to 100. The city must set a threshold: anyone above the threshold is released on bail.

**Three possible approaches:**

| Approach | Rule | Consequence |
|----------|------|-------------|
| **Same threshold** | Everyone with score > 50 is released | More Group B detained (because their scores are lower on average) |
| **Equal release rates** | Adjust thresholds so 60% of each group is released | More Group A members who would have skipped court are released; public safety concern |
| **Equal false detention rates** | Ensure the same percentage of people who *would have* shown up are wrongly detained in both groups | Different release rates; Group B gets a lower threshold, which some call "preferential treatment" |

**Your task:**

1. For **each approach**, identify who is harmed and how. Be specific: don't just say "it's unfair." Describe a person who suffers under each rule.

2. Choose the approach you think is least bad. Write at least 3-4 sentences defending your choice. You must address at least one objection someone could raise against your choice.

3. The reasons Group B has a lower show-up rate (transport, jobs, childcare) are not about criminal intent. Does this change which approach you'd pick? Should the algorithm account for *why* someone is risky, or just *whether* they are? Explain in 3–4 sentences.
