# DATASCI 101: group project random assignment

This notebook assigns students to groups of 4-5 for the final project.  
**Groups 01-06** have self-formed cores. **Groups 07-09** are fully random.  
Groups 05 and 06 were topped up with random students to reach 5.  
Random seed is `185` for reproducibility.

> **Note:** email addresses follow the `firstname.lastname@emory.edu` pattern and should be verified.

```python
import random
import pandas as pd
from IPython.display import display, Markdown

def make_email(name):
    """Generate email: firstname.lastname@emory.edu."""
    return ".".join(name.lower().split()) + "@emory.edu"

# --- Pre-formed groups ---
group_01 = ["Student 11", "Student 23", "Student 27"]
group_02 = ["Student 05", "Student 16", "Student 22", "Student 37", "Student 39"]
group_03 = ["Student 02", "Student 12", "Student 26", "Student 32"]
group_04 = ["Student 13", "Student 18", "Student 28"]
group_05 = ["Student 04", "Student 10", "Student 29", "Student 35"]  # needs 1 more
group_06 = ["Student 21", "Student 24"]                                         # needs 3 more

pre_formed = [group_01, group_02, group_03, group_04, group_05, group_06]
pre_formed_names = {name for g in pre_formed for name in g}

# --- Full student roster (alphabetical) ---
# Excludes: instructor, TAs, and graders
all_students = [
    "Student 01",
    "Student 02",
    "Student 03",
    "Student 04",
    "Student 05",
    "Student 06",
    "Student 07",
    "Student 08",
    "Student 09",
    "Student 10",
    "Student 11",
    "Student 12",
    "Student 13",
    "Student 14",
    "Student 15",
    "Student 16",
    "Student 17",
    "Student 18",
    "Student 19",
    "Student 20",
    "Student 21",
    "Student 22",
    "Student 23",
    "Student 24",
    "Student 25",
    "Student 26",
    "Student 27",
    "Student 28",
    "Student 29",
    "Student 30",
    "Student 31",
    "Student 32",
    "Student 33",
    "Student 34",
    "Student 35",
    "Student 36",
    "Student 37",
    "Student 38",
    "Student 39",
    "Student 40"
]

# --- Randomly assign remaining students ---
remaining = [s for s in all_students if s not in pre_formed_names]

random.seed(185)
shuffled = remaining.copy()
random.shuffle(shuffled)

# Top up groups 05 and 06 from the shuffled pool
group_05.append(shuffled[0])           # +1 -> 5 members
group_06.extend(shuffled[1:4])         # +3 -> 5 members
purely_random = shuffled[4:]           # 15 remaining -> 3 groups of 5

new_groups = [purely_random[i:i + 5] for i in range(0, len(purely_random), 5)]

# Combine all groups
all_groups = pre_formed + new_groups

print(f"Total students: {len(all_students)}")
print(f"Self-formed cores: {len(pre_formed_names)} (Groups 01-06)")
print(f"Random pool: {len(remaining)} (topped up 05-06, then Groups 07-09)")
print(f"Total groups: {len(all_groups)}")
```

```python
# Display all groups
n_pre = len(pre_formed)
for i, group in enumerate(all_groups, 1):
    if i <= n_pre:
        label = "(self-formed)"
    else:
        label = "(random)"
    display(Markdown(f"### Group {i:02d} {label}"))
    df = pd.DataFrame(
        [{"Name": s, "Email": make_email(s)} for s in sorted(group)]
    )
    df.index += 1
    display(df)

# Summary table
display(Markdown("---\n### Summary"))
rows = []
for i, group in enumerate(all_groups, 1):
    formed = "self-formed" if i <= n_pre else "random"
    for s in group:
        rows.append({"Group": f"{i:02d}", "Name": s, "Email": make_email(s), "Type": formed})
summary = pd.DataFrame(rows).sort_values("Name").reset_index(drop=True)
summary.index += 1
display(summary)
```
